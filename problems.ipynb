```python
import numpy as np

# ---------------------------------------------------------------------
# Helper functions
# ---------------------------------------------------------------------

def rotr(x, n):
    """
    Rotate right: return the 32-bit value of x rotated right by n bits.
    Bits shifted off the right end reappear on the left.
    """
    x = np.uint32(x)
    return np.uint32((x >> n) | (x << (32 - n)))

def shr(x, n):
    """
    Logical right shift: shift right by n bits, filling with zeros.
    """
    x = np.uint32(x)
    return np.uint32(x >> n)

# ---------------------------------------------------------------------
# Core Boolean and Bitwise Functions (FIPS 180-4, page 10)
# ---------------------------------------------------------------------

def Parity(x, y, z):
    """
    Parity(x, y, z) = x XOR y XOR z
    Bitwise XOR of three 32-bit words.
    """
    return np.uint32(x) ^ np.uint32(y) ^ np.uint32(z)

def Ch(x, y, z):
    """
    Ch(x, y, z) = (x AND y) XOR ((NOT x) AND z)
    'Choose' function: each bit of x selects the corresponding bit from y or z.
    """
    x, y, z = np.uint32(x), np.uint32(y), np.uint32(z)
    return np.uint32((x & y) ^ (~x & z))

def Maj(x, y, z):
    """
    Maj(x, y, z) = (x AND y) XOR (x AND z) XOR (y AND z)
    'Majority' function: for each bit, output 1 if at least two of x, y, z bits are 1.
    """
    x, y, z = np.uint32(x), np.uint32(y), np.uint32(z)
    return np.uint32((x & y) ^ (x & z) ^ (y & z))

# ---------------------------------------------------------------------
# Sigma and sigma functions
# ---------------------------------------------------------------------

def Sigma0(x):
    """Σ₀(x) = ROTR²(x) XOR ROTR¹³(x) XOR ROTR²²(x)"""
    x = np.uint32(x)
    return rotr(x, 2) ^ rotr(x, 13) ^ rotr(x, 22)

def Sigma1(x):
    """Σ₁(x) = ROTR⁶(x) XOR ROTR¹¹(x) XOR ROTR²⁵(x)"""
    x = np.uint32(x)
    return rotr(x, 6) ^ rotr(x, 11) ^ rotr(x, 25)

def sigma0(x):
    """σ₀(x) = ROTR⁷(x) XOR ROTR¹⁸(x) XOR SHR³(x)"""
    x = np.uint32(x)
    return rotr(x, 7) ^ rotr(x, 18) ^ shr(x, 3)

def sigma1(x):
    """σ₁(x) = ROTR¹⁷(x) XOR ROTR¹⁹(x) XOR SHR¹⁰(x)"""
    x = np.uint32(x)
    return rotr(x, 17) ^ rotr(x, 19) ^ shr(x, 10)
```

```python
def block_parse(msg):
    """
    Generator function that parses a message into 512-bit (64-byte) blocks
    according to FIPS 180-4 sections 5.1.1 and 5.2.1.
    
    The message is padded with:
    - A single '1' bit (0x80 byte when followed by zeros)
    - Zero padding
    - A 64-bit length field (in bits) at the end
    
    Args:
        msg: bytes object containing the message to be parsed
        
    Yields:
        512-bit (64-byte) blocks as bytes objects
    """
    # Get the length of the message in bits
    msg_len_bits = len(msg) * 8
    
    # Start with the original message
    padded = bytearray(msg)
    
    # Step 1: Append the bit '1' (represented as 0x80)
    padded.append(0x80)
    
    # Step 2: Append zero bytes until we are 8 bytes (64 bits) from a 512-bit boundary
    # 512 bits = 64 bytes, so we need length % 64 == 56 (leaving 8 bytes for length field)
    while len(padded) % 64 != 56:
        padded.append(0x00)
    
    # Step 3: Append the original message length as a 64-bit big-endian integer
    padded.extend(msg_len_bits.to_bytes(8, byteorder='big'))
    
    # Step 4: Yield 512-bit blocks
    for i in range(0, len(padded), 64):
        yield bytes(padded[i:i+64])
```

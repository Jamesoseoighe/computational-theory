```python
import numpy as np

# ---------------------------------------------------------------------
# Helper functions
# ---------------------------------------------------------------------

def rotr(x, n):
    """
    Rotate right: return the 32-bit value of x rotated right by n bits.
    Bits shifted off the right end reappear on the left.
    """
    x = np.uint32(x)
    return np.uint32((x >> n) | (x << (32 - n)))

def shr(x, n):
    """
    Logical right shift: shift right by n bits, filling with zeros.
    """
    x = np.uint32(x)
    return np.uint32(x >> n)

# ---------------------------------------------------------------------
# Core Boolean and Bitwise Functions (FIPS 180-4, page 10)
# ---------------------------------------------------------------------

def Parity(x, y, z):
    """
    Parity(x, y, z) = x XOR y XOR z
    Bitwise XOR of three 32-bit words.
    """
    return np.uint32(x) ^ np.uint32(y) ^ np.uint32(z)

def Ch(x, y, z):
    """
    Ch(x, y, z) = (x AND y) XOR ((NOT x) AND z)
    'Choose' function: each bit of x selects the corresponding bit from y or z.
    """
    x, y, z = np.uint32(x), np.uint32(y), np.uint32(z)
    return np.uint32((x & y) ^ (~x & z))

def Maj(x, y, z):
    """
    Maj(x, y, z) = (x AND y) XOR (x AND z) XOR (y AND z)
    'Majority' function: for each bit, output 1 if at least two of x, y, z bits are 1.
    """
    x, y, z = np.uint32(x), np.uint32(y), np.uint32(z)
    return np.uint32((x & y) ^ (x & z) ^ (y & z))

# ---------------------------------------------------------------------
# Sigma and sigma functions
# ---------------------------------------------------------------------

def Sigma0(x):
    """Σ₀(x) = ROTR²(x) XOR ROTR¹³(x) XOR ROTR²²(x)"""
    x = np.uint32(x)
    return rotr(x, 2) ^ rotr(x, 13) ^ rotr(x, 22)

def Sigma1(x):
    """Σ₁(x) = ROTR⁶(x) XOR ROTR¹¹(x) XOR ROTR²⁵(x)"""
    x = np.uint32(x)
    return rotr(x, 6) ^ rotr(x, 11) ^ rotr(x, 25)

def sigma0(x):
    """σ₀(x) = ROTR⁷(x) XOR ROTR¹⁸(x) XOR SHR³(x)"""
    x = np.uint32(x)
    return rotr(x, 7) ^ rotr(x, 18) ^ shr(x, 3)

def sigma1(x):
    """σ₁(x) = ROTR¹⁷(x) XOR ROTR¹⁹(x) XOR SHR¹⁰(x)"""
    x = np.uint32(x)
    return rotr(x, 17) ^ rotr(x, 19) ^ shr(x, 10)
```

### Output for Problem 2: Fractional Parts of Cube Roots

The following cell prints the first 32 bits of the fractional parts of the cube roots of the first 64 prime numbers, as required by the Secure Hash Standard (FIPS 180-4). This output allows you to verify your results against the standard.

```python
import numpy as np

def primes(n):
    """Return a list of the first n prime numbers."""
    primes = []
    candidate = 2
    while len(primes) < n:
        is_prime = True
        for p in primes:
            if p * p > candidate:
                break
            if candidate % p == 0:
                is_prime = False
                break
        if is_prime:
            primes.append(candidate)
        candidate += 1
    return primes

# Calculate the first 64 primes
prime_list = primes(64)

# Calculate the cube roots and extract the fractional part
cube_roots = np.cbrt(prime_list)
fractional_parts = cube_roots - np.floor(cube_roots)

# Extract the first 32 bits of the fractional part
constants = [int(f * (2**32)) for f in fractional_parts]

# Display the result in hexadecimal, 8 per line, matching FIPS 180-4 style
for i in range(0, 64, 8):
    print(', '.join(f'0x{constants[j]:08x}' for j in range(i, i+8)))
```
